```python
import pandas as pd

reviews = pd.DataFrame([["Great game, terrible teammates",False,1],["I love this game",True,0],["Best matchmaking ever... lost ten games",False,1],["Fun with friends",True,0],["Servers are laggy",False,0],["Very satisfying gameplay",True,0]], columns=['review', 'recommended', 'irony_cue'])
reviews.head()
```

```python
reviews.groupby('recommended').size()
```

```python
reviews['irony_cue'].value_counts().plot(kind='bar', title='Irony cue frequency', rot=0)
```
